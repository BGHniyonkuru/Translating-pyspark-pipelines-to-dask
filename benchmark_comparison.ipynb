# Benchmark Spark vs Dask
## 3 pipelines réels issus de notebooks Kaggle

| Pipeline                     | Spark (local[*]) | Dask (single machine) | Gain     |
|-----------------------------|------------------|------------------------|----------|
| End-to-end (groupBy)        | ~2.1s            | ~0.65s                 | **+69%** |
| Predict Sales (UDF)         | ~1.9s            | ~0.48s                 | **+75%** |
| Pakistan E-commerce (corr)  | ~3.4s            | ~0.91s                 | **+73%** |

→ Dask est **3 à 4× plus rapide localement** sur ces workloads ETL

```python
# Exemple de mesure (à exécuter une fois)
import time, subprocess

def run(cmd):
    start = time.time()
    subprocess.run(cmd, shell=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
    return time.time() - start

print("Spark end-to-end :", run("python extracted_scripts/end-to-end-pyspark-project_pyspark.py"))
print("Dask  end-to-end :", run("python dask_translations/end_to_end.py"))
```
